## Movies Storytelling

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/movies.csv")
```

```python
RED = "#E53935"
BLUE = "#1E88E5"
GRAY_LIGHT = "#EAEAEA"
GRAY_TEXT = "#37474F"
BACKGROUND = "#FFFFFF"


df["release_date"] = pd.to_datetime(df["release_date"])
df["release_year"] = df["release_date"].dt.year
```

```python
df.head()
```

### Betyg, Vilka filmer når toppen av publikens betyg?

```python
top10 = (
    df.sort_values("vote_average", ascending=False)
    .head(10)
    .sort_values("vote_average")
)
colors_1 = [GRAY_LIGHT] * (len(top10) - 1) + [RED]

fig, ax1 = plt.subplots(figsize=(9, 5))
fig.patch.set_facecolor(BACKGROUND)
ax1.set_facecolor(BACKGROUND)

bars = ax1.barh(
    top10["title"], top10["vote_average"], color=colors_1, height=0.6
)

for bar in bars:
    val = bar.get_width()
    ax1.text(
        val + 0.05,
        bar.get_y() + bar.get_height() / 2,
        f"{val:.2f}",
        va="center",
        color=GRAY_TEXT,
        fontsize=9,
    )

ax1.spines[["top", "right", "left"]].set_visible(False)
ax1.spines["bottom"].set_color(GRAY_LIGHT)
ax1.tick_params(axis="both", colors=GRAY_TEXT, length=0)
ax1.set_xlim(0, 10)


ax1.set_title(
    "1. Vilka filmer når toppen av publikens betyg?\n",
    loc="left",
    color=GRAY_TEXT,
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax1.set_xlabel("Genomsnittligt betyg (0–10)", color=GRAY_TEXT, fontsize=9)

fig.tight_layout()
fig.savefig("1_top10_filmer.png", dpi=150, bbox_inches="tight")
plt.show()
```

### POPULARITET VS BETYG, Är de högst betygsatta filmerna också de mest populära?

"Högst betyg innebär inte alltid högst popularitet i datan"

```python
fig, ax2 = plt.subplots(figsize=(9, 5))
fig.patch.set_facecolor(BACKGROUND)
ax2.set_facecolor(BACKGROUND)

ax2.scatter(
    df["vote_average"],
    df["popularity"],
    color=BLUE,
    alpha=0.6,
    edgecolors="none",
)

ax2.spines[["top", "right"]].set_visible(False)
ax2.spines[["left", "bottom"]].set_color(GRAY_LIGHT)
ax2.tick_params(axis="both", colors=GRAY_TEXT)


ax2.set_title(
    "2. Är de högst betygsatta filmerna också de mest populära?\n"
    ,
    loc="left",
    color=GRAY_TEXT,
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax2.set_xlabel("Genomsnittligt betyg", color=GRAY_TEXT, fontsize=9)
ax2.set_ylabel("Popularitetspoäng", color=GRAY_TEXT, fontsize=9)

fig.tight_layout()
fig.savefig("2_betyg_vs_popularitet.png", dpi=150, bbox_inches="tight")
plt.show()
```

### UTGIVNINGSÅR, När producerades filmerna på topplistan?

```python
fig, ax3 = plt.subplots(figsize=(9, 5))
fig.patch.set_facecolor(BACKGROUND)
ax3.set_facecolor(BACKGROUND)

ax3.scatter(
    df["release_year"],
    df["vote_average"],
    color=RED,
    alpha=0.5,
    s=30,
)

ax3.spines[["top", "right"]].set_visible(False)
ax3.spines[["left", "bottom"]].set_color(GRAY_LIGHT)
ax3.tick_params(axis="both", colors=GRAY_TEXT)

ax3.set_title(
    "3. När producerades filmerna på topplistan?\n",
    loc="left",
    color=GRAY_TEXT,
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax3.set_xlabel("Utgivningsår", color=BLUE, fontsize=9)
ax3.set_ylabel("Genomsnittligt betyg", color=BLUE, fontsize=9)

fig.tight_layout()
fig.savefig("3_utgivningsar_betyg.png", dpi=150, bbox_inches="tight")
plt.show()
```
